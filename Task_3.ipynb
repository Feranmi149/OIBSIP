```python
df = pd.read_csv('The Titanic dataset.csv', header=1)
df.rename(columns={'Unnamed: 3': 'name'}, inplace=True)
df.head()
```

```python
print("Shape:", df.shape)
print("\nNulls per column:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nData types:\n", df.dtypes)
```

Data Quality Report (Before Cleaning): 1,301 rows, 10 columns. Missing values found in gender (1), age (257), family (2), fare (2), and embarked (6). One duplicate row identified. Age and fare are incorrectly typed as object/text due to placeholder values like "?" and "**" representing missing data.

```python
import numpy as np
df['age'] = df['age'].replace('?', np.nan)
df['fare'] = df['fare'].replace('**', np.nan)
```

```python
df['age'] = pd.to_numeric(df['age'])
df['fare'] = pd.to_numeric(df['fare'])
```

```python
df.isnull().sum()
```

```python
df['gender'] = df['gender'].fillna(df['gender'].mode()[0])
```

Justification: Gender is categorical with only 1 missing value; filling with the most common value (mode) is reasonable and has negligible impact on the dataset.

```python
df['age'] = df['age'].fillna(df['age'].median())
```

Justification: Age has a large number of missing values (258). Median is used instead of mean because age can be skewed by outliers (e.g., infants or elderly passengers), and median is more robust to that skew.

```python
df['family'] = df['family'].fillna(df['family'].mode()[0])
```

Justification: Family size is a count value; only 2 records are missing, so filling with the most common family size (mode) is a safe, low-impact choice.

```python
df['fare'] = df['fare'].fillna(df['fare'].median())
```

Justification: Fare is monetary and can have outliers (luxury cabins vs. steerage); median avoids skew from extreme high fares better than mean.

```python
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

Embarkation port is categorical with only 6 missing values; mode imputation is standard practice here since it's low-impact and preserves the most common category.

```python
df.isnull().sum()
```

```python
print("Duplicates before:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after:", df.duplicated().sum())
print("New shape:", df.shape)
```

Duplicate Removal: 1 duplicate row was identified and removed (the repeated "Mr. Anthony" record). Row count reduced from 1301 to 1300.

```python
df['gender'].unique()
```

```python
df['embarked'].unique()
```

Gender (male/female) and embarked (S, C, Q) columns were already consistently formatted so no changes needed.

```python
df['date'] = pd.to_datetime(df['date'], format='%d-%b-%y')
df.dtypes
```

```python
df['sn'] = df['sn'].astype(str)
df.dtypes
```

Data Type Correction: All columns now have appropriate types, sn converted to string (as an identifier, not a quantity), age/family/fare as float, date as proper datetime, and categorical columns (gender, embarked) remain as text.

```python
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers_age = df[(df['age'] < lower) | (df['age'] > upper)]
print("Age outliers:", outliers_age.shape[0])
print("Valid range:", lower, "to", upper)
```

```python
Q1 = df['fare'].quantile(0.25)
Q3 = df['fare'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers_fare = df[(df['fare'] < lower) | (df['fare'] > upper)]
print("Fare outliers:", outliers_fare.shape[0])
print("Valid range:", lower, "to", upper)
```

Outlier Detection (IQR method): Using the IQR method, 99 age outliers were identified (valid range: 2.5–54.5) and 171 fare outliers were identified (valid range: -27.17 to 66.34). Reviewing the data, age outliers represent genuinely older passengers rather than data entry errors, ship passengers naturally span a wide age range. Fare outliers largely reflect higher-class ticket prices (first-class fares were legitimately much higher than steerage), which is expected given how fare correlates with passenger class rather than being erroneous. Both sets of outliers were retained rather than capped or removed, since removing them would distort the natural variation in passenger age and ticket pricing that the dataset is meant to reflect.
